```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()

llm= ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
with open("blood_work.txt", "r") as f:
    blood_report =f.read()

print(blood_report[:200])
```

```python
extraction_prompt=f"""
You  are a medical data extraction assistant.
From the blood report below ,extract all test value and classify each one as HIGH,LOW,or NORMAL 
based on the reference ranges provided in report.

Format your response as:
 -Test Name: value|Status:HIGH/LOW/NORMAL |Reference: range

 Blood Report:
 {blood_report}
    """
extraction_response = llm.invoke(extraction_prompt)
extraction_value=extraction_response.text
print(extraction_value)
```
