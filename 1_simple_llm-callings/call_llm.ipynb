```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm=ChatGoogleGenerativeAI(model="gemma-4-31b-it")

response=llm.invoke("how any moons in jupiter?")
print(response.text)
```

```python
llm=ChatGoogleGenerativeAI(model="gemma-4-31b-it")

response=llm.invoke([
    ["system","you are a helpful assistant that answer in short one line"],
    ["user","how many moons that jupiter have"]
    
])
print(response.text)
```
